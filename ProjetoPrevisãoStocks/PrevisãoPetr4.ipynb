```python
import yfinance as yf

# Ticker da PETR4 na B3 
petr4 = yf.Ticker("PETR4.SA")

# Histórico max de preços
dados = petr4.history(period="max")

# Visualizar as 5 primeiras linhas
dados.head()
```

```python
del dados["Dividends"]
del dados["Stock Splits"]
```

```python
dados["Amanha"] = dados["Close"].shift(-1)
```

```python
dados["Objetivo"] = (dados["Amanha"] > dados["Close"]).astype(int)
```

```python
from sklearn.ensemble import RandomForestClassifier

# Começando a treinar modelo de Machine Learning (ML)
model = RandomForestClassifier(n_estimators=100, min_samples_leaf=100, random_state=1)

# Separando os dados em treino e teste
# Aqui, os últimos 100 registros serão usados como teste
train = dados.iloc[:-100]   # Todas as linhas menos as últimas 100
test = dados.iloc[-100:]    # As últimas 100 linhas do conjunto de dados

predictors = ["Close", "Volume", "Open", "High", "Low"]

# Treinando o modelo com os dados de treino
# treino[Predictors] são os dados de entrada (X)
# treino["Objetivo"] é a variável alvo (y), que o modelo deve aprender a prever
model.fit(train[predictors], train["Objetivo"])
```

```python
#Saber o quão bom é o modelo:
from sklearn.metrics import precision_score

#Quantos % que o modelo preveu que iria subir o preço, realmente subiu

preds = model.predict(test[predictors])
```

```python
import pandas as pd

preds = pd.Series(preds, index=test.index)
```

```python
precision_score(test["Objetivo"], preds) 
#Comparando o que a máquina disse com resultados reais, apenas 44%
```

```python
#Backtesting - Testando com varios anos
def predict(train, test, predictors, model):
    model.fit(train[predictors], train["Objetivo"])
    preds = model.predict(test[predictors])
    preds = pd.Series(preds, index=test.index, name="Predictions")
    combined = pd.concat([test["Objetivo"], preds], axis=1)
    return combined
```

```python
def backtest(data, model, predictors, start=2500, step=250):
    #pegue 10 anos de data e treine o seu primeiro modelo, treinaremos um modelo por 1 ano e iremos prever o proximo
    all_predictions = []
    
    for i in range(start, data.shape[0], step):
        train = data.iloc[0:i].copy()
        test = data.iloc[i:(i+step)].copy()
        predictions = predict(train, test, predictors, model)
        all_predictions.append(predictions)
    
    return pd.concat(all_predictions)
    
```

```python
predictions = backtest(dados, model, predictors)
```

```python
predictions["Predictions"].value_counts()
```

```python
precision_score(predictions["Objetivo"], predictions["Predictions"])
```

```python
horizons = [2,5,60,250,1000]
#2 dias/5 dias/60 dias/250 dias/1000 dias
new_predictors = []

#Ver os ultimos dias de cada horizonte, adicionando colunas de dados
for horizon in horizons:
    rolling_avarages = dados.rolling(horizon).mean()
    
    ratio_column = f"Close_ratio_{horizon}"
    dados[ratio_column] = dados["Close"] / rolling_avarages["Close"]
    
    trend_column = f"Trend_{horizon}"
    dados[trend_column] = dados.shift(1).rolling(horizon).sum()["Objetivo"]
    
    new_predictors += [ratio_column, trend_column]
    
```

```python
dados = dados.dropna()
```

```python
model = RandomForestClassifier(n_estimators=200,min_samples_split=50, random_state=1)
```

```python
def predict(train, test, predictors, model):
    model.fit(train[predictors], train["Objetivo"])
    preds = model.predict_proba(test[predictors]) [:,1]
    #Vai retornar a probabilidade de elevar ou abaixar o preço
    preds[preds >= .6] = 1
    preds[preds < .6] = 0
    #aumentar a chance do preço subir, fazendo um buy and hold
    preds = pd.Series(preds, index=test.index, name="Predictions")
    combined = pd.concat([test["Objetivo"], preds], axis=1)
    return combined
```

```python
predictions = backtest(dados, model, new_predictors)
```

```python
predictions["Predictions"].value_counts()
```

```python
precision_score(predictions["Objetivo"], predictions["Predictions"])
```
